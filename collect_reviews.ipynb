```python
from google_play_scraper import reviews, Sort
import pandas as pd

# pull the 50,000 most recent English reviews of Duolingo from the US Google Play store
result, _ = reviews(
    "com.duolingo",
    lang="en",
    country="us",
    sort=Sort.NEWEST,
    count=50000,
)

df = pd.DataFrame(result)

# keep only the columns I need, and drop usernames/profile pics for privacy
df = df[["reviewId", "content", "score", "thumbsUpCount", "reviewCreatedVersion", "at"]]

df.to_csv("data/duolingo_reviews.csv", index=False)

print(df.shape)
print("Oldest review:", df["at"].min())
print("Newest review:", df["at"].max())
df.head()
```

```python
print("Rows:", len(df))
print("\nStar ratings:")
print(df["score"].value_counts().sort_index())
print("\n% that are 1-2 stars:", round((df["score"] <= 2).mean() * 100, 1))
print("Number of app versions:", df["reviewCreatedVersion"].nunique())
print("Missing review text:", df["content"].isna().sum())
```
